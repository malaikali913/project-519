```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# 1. Tolerance Levels (epsilon)
# =====================================================
eps = np.array([0.01, 0.05, 0.10, 0.20, 0.30])

# =====================================================
# 2. Lyapunov Stability (Energy deviation)
#    You should replace these values with real simulation
# =====================================================
Lyap = np.array([0.02, 0.05, 0.10, 0.22, 0.38])  
# Smaller = MORE stable

# =====================================================
# 3. MCDM/TOPSIS Stability Scores
#    These are example values — replace with your own if needed
# =====================================================
Topsis = np.array([0.82, 0.74, 0.62, 0.45, 0.33])  
# Larger = MORE stable

# =====================================================
# 4. Normalize Lyapunov to 0–1 RANGE for fair comparison
#    (Since Lyapunov is a cost)
# =====================================================
Lyap_norm = 1 - (Lyap / Lyap.max())

# =====================================================
# 5. Plot Both Curves
# =====================================================
plt.figure(figsize=(10,6))

plt.plot(eps, Lyap_norm, marker='o', label="Lyapunov Stability (normalized)")
plt.plot(eps, Topsis, marker='o', label="TOPSIS Stability Score")

plt.title("Comparison of Lyapunov vs MCDM-TOPSIS Stability")
plt.xlabel("Tolerance ε")
plt.ylabel("Stability (0 to 1)")
plt.grid(True)
plt.legend()
plt.show()

# =====================================================
# 6. Print Interpretation of Relationship
# =====================================================

print("=== Stability Relationship Between Lyapunov and TOPSIS ===")

for i in range(len(eps)):
    print(f"ε={eps[i]} | Lyapunov={Lyap_norm[i]:.3f} | TOPSIS={Topsis[i]:.3f}")

corr = np.corrcoef(Lyap_norm, Topsis)[0,1]
print("\nCorrelation between Lyapunov and TOPSIS =", round(corr, 3))

if corr > 0.7:
    print("→ Strong agreement: Both methods show similar stability trend.")
elif corr > 0.4:
    print("→ Moderate agreement: TOPSIS adds more detail than Lyapunov.")
else:
    print("→ Weak agreement: TOPSIS captures more information than Lyapunov.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.fft import fft, ifft, fftfreq
from scipy.sparse.linalg import eigs
from scipy.linalg import solve_banded

# =============================================================
# 1. Analytical soliton
# =============================================================
def analytical_soliton(x, w):
    return np.sqrt(2*w) * np.cosh(np.sqrt(w)*x)**(-1)

# =============================================================
# 2. Numerical soliton using Newton iteration
#     Solve: -phi_xx + w phi - phi^3 = 0
# =============================================================
def numerical_soliton_newton(x, w, tol=1e-10, max_iter=50):
    h = x[1] - x[0]
    phi = analytical_soliton(x, w) * 0.9   # initial guess

    for it in range(max_iter):
        # Residual: F = -phi'' + w phi - phi^3
        phi_xx = (np.roll(phi,-1) - 2*phi + np.roll(phi,1)) / h**2
        F = -phi_xx + w*phi - phi**3

        # Jacobian diagonal for Newton: J = -d2/dx2 + w - 3phi^2
        J_main = (2/h**2 + w - 3*phi**2)
        J_upper = (-1/h**2) * np.ones_like(phi)
        J_lower = (-1/h**2) * np.ones_like(phi)

        # Banded matrix form (superdiagonal, diagonal, subdiagonal)
        ab = np.zeros((3, len(phi)))
        ab[0,1:] = J_upper[:-1]
        ab[1,:] = J_main
        ab[2,:-1] = J_lower[1:]

        delta = solve_banded((1,1), ab, -F)
        phi = phi + delta

        if np.linalg.norm(delta) < tol:
            break
    
    return phi

# =============================================================
# 3. Linear Stability Eigenvalue Test (JLW = λW)
# =============================================================
def linear_stability(phi, w, x):
    N = len(phi)
    h = x[1] - x[0]

    # Operators L+ and L-
    Lp = -np.diag(np.ones(N-1),-1) + 2*np.diag(np.ones(N)) - np.diag(np.ones(N-1),1)
    Lp = Lp/h**2 + np.diag(w - 3*phi**2)

    Lm = -np.diag(np.ones(N-1),-1) + 2*np.diag(np.ones(N)) - np.diag(np.ones(N-1),1)
    Lm = Lm/h**2 + np.diag(w - phi**2)

    # Block matrix JL
    J = np.block([[np.zeros((N,N)), np.eye(N)],
                  [-np.eye(N), np.zeros((N,N))]])

    L = np.block([[Lp, np.zeros((N,N))],
                  [np.zeros((N,N)), Lm]])

    JL = J @ L

    # Compute several eigenvalues near origin
    vals, _ = eigs(JL, k=6, sigma=0)  
    return vals

# =============================================================
# 4. Conservation Laws
# =============================================================
def mass(psi, dx):
    return np.sum(np.abs(psi)**2) * dx

def energy(psi, dx):
    psi_x = (np.roll(psi,-1) - np.roll(psi,1)) / (2*dx)
    return np.sum(np.abs(psi_x)**2 - 0.5*np.abs(psi)**4) * dx

# =============================================================
# 5. Split-Step Fourier Method (SSFM)
# =============================================================
def ssfm(psi0, x, dt, T, w):
    N = len(x)
    dx = x[1] - x[0]

    k = fftfreq(N, dx) * 2*np.pi
    psi = psi0.copy()

    nsteps = int(T/dt)

    mlist = []
    elist = []

    for _ in range(nsteps):
        # Half linear step
        psi_hat = fft(psi)
        psi_hat *= np.exp(-1j*k**2*(dt/2))
        psi = ifft(psi_hat)

        # Nonlinear step
        psi *= np.exp(1j * np.abs(psi)**2 * dt)

        # Half linear step
        psi_hat = fft(psi)
        psi_hat *= np.exp(-1j*k**2*(dt/2))
        psi = ifft(psi_hat)

        # Record conservation
        mlist.append(mass(psi, dx))
        elist.append(energy(psi, dx))
    
    return psi, np.array(mlist), np.array(elist)

# =============================================================
# 6. MAIN VERIFICATION RUN
# =============================================================
L = 20
N = 600
x = np.linspace(-L, L, N)
w = 1.0

# Analytical soliton
phi_exact = analytical_soliton(x, w)

# Numerical soliton
phi_num = numerical_soliton_newton(x, w)

# --- Verification 1: Numerical vs Analytical
err_L2 = np.linalg.norm(phi_num - phi_exact)
print("L2 error between numerical and analytical soliton =", err_L2)

# --- Verification 2: Eigenvalue Check
vals = linear_stability(phi_num, w, x)
print("\nEigenvalues near origin:")
print(vals)
print("Max real part =", np.max(vals.real))

# --- Verification 3: Conservation Laws
psi0 = phi_num
psi_final, M, E = ssfm(psi0, x, dt=0.005, T=2.0, w=w)

print("\nConservation:")
print("Initial Mass =", M[0], "Final Mass =", M[-1])
print("Initial Energy =", E[0], "Final Energy =", E[-1])

# --- Verification 4: Plot Conservation
plt.figure(figsize=(10,4))
plt.plot(M, label="Mass Q(t)")
plt.plot(E, label="Energy E(t)")
plt.title("Conservation of Mass and Energy")
plt.legend()
plt.grid(True)
plt.show()

# --- Time Evolution Stability Error
stab_error = np.linalg.norm(np.abs(psi_final) - phi_num)
print("\nTime evolution stability error =", stab_error)
```
